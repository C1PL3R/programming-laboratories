## Циклічні алгоритми

Циклічний алгоритм — це алгоритм, у якому певна послідовність команд виконується декілька разів. У Python основними конструкціями для організації циклів є `for` та `while`.

Цикл `for` використовують для послідовного перебору елементів відомої послідовності або діапазону:

```python
for number in range(1, 11):
    print(number)

```

Для генерування числових послідовностей використовується функція `range()`:

```python
range(stop)
range(start, stop)
range(start, stop, step)

```

Верхня межа `stop` до послідовності не входить. Наприклад, `range(1, 6)` відповідає числам: `1, 2, 3, 4, 5`. Для перебору всіх натуральних чисел від `1` до `N` включно використовують `range(1, N + 1)`.


## Цикл `while`

Цикл `while` використовується тоді, коли кількість повторень наперед невідома, а зациклення регулюється певною логічною умовою.

Загальна структура:

```python
while condition:
    instructions

```

Наприклад:

```python
number = 100

while number > 0:
    print(number)
    number //= 10

```

Після кожного виконання тіла циклу стан змінних повинен змінюватися так, щоб умова циклу зрештою стала хибною (для уникнення безкінечного циклу).

## Послідовне опрацювання цифр числа

За допомогою цілочисельного ділення (`//`) та остачі від ділення (`%`) можна розкладати ціле число на цифри справа наліво (від молодших розрядів до старших).

Загальна схема перебору:

```python
while number > 0:
    digit = number % 10

    # опрацювання чергової цифри (digit)

    number //= 10

```

Ця схема є базовою для реалізації таких типових операцій:

* обчислення суми або добутку цифр;
* знаходження кількості цифр у числі;
* пошук екстремумів (найбільшої або найменшої цифри);
* інвертування (реверсу) числа.

Якщо значення початкового числа потрібне після завершення циклу, його слід попередньо зберегти в окремій змінній: `original = number`.

## Формування нового числа з цифр

За допомогою арифметичних операцій можна конструювати нове число шляхом послідовного додавання цифр у молодші розряди.

Якщо поточне сформоване число зберігається в `result`, а нова цифра — в `digit`, то конструкція:

```python
result = result * 10 + digit

```

зсуває наявні цифри `result` на один розряд ліворуч і дописує `digit` у кінець.


## Математичні методи роботи з цифрами та розрядами

Опрацьовувати розряди цілих чисел можна не лише циклічним перебором або конвертацією у рядкові типи даних, а й суто математичними методами.

### 1. Математичне визначення кількості цифр

Кількість цифр $k$ у натуральному числі $n$ можна обчислити за допомогою десяткового логарифма $\log_{10}$:

$$k = \lfloor\log_{10}(n)\rfloor + 1$$

Приклад реалізації на Python:

```python
import math

count = math.floor(math.log10(number)) + 1

```

### 2. Математичне отримання цифри за позицією

Щоб виділити $pos$-ту цифру (починаючи з 1, зліва направо) із числа $n$, яке має загалом $k$ цифр, виконують зсув розрядів за допомогою ступенів 10:

$$\text{digit} = \left\lfloor \frac{n}{10^{k - pos}} \right\rfloor \bmod 10$$

```python
shift = total_digits - pos
digit = (number // (10 ** shift)) % 10

```

## Математичні характеристики та числові класи

В теорії чисел виділяють різноманітні класи чисел за їхніми структурними або подільними властивостями.

### Числа Армстронга (нарцисичні числа)

Числом Армстронга називається натуральне число $N$, яке дорівнює сумі своїх цифр, піднесених до степеня, що дорівнює їхній загальній кількості $k$:

$$N = \sum_{i=1}^{k} d_i^k$$

де $d_i$ — окремі цифри числа.

### Простота числа

Натуральне число $n > 1$ називається простим, якщо воно має лише два дільники: $1$ та $n$.

Для перевірки простоти числа не потрібно перебирати всі значення від $2$ до $n-1$. Якщо число складене, воно обов'язково має принаймні один дільник, що не перевищує $\sqrt{n}$. Тому верхню межу шуканого дільника обмежують значенням $\lfloor\sqrt{n}\rfloor$:

```python
import math

def is_prime(number: int) -> bool:
    if number < 2:
        return False
    for d in range(2, math.isqrt(number) + 1):
        if number % d == 0:
            return False
    return True

```

## Алгоритми теорії подільності

### Найбільший спільний дільник (НСД)

Найбільшим спільним дільником двох цілих чисел називається найбільше натуральне число, на яке обидва числа діляться без остачі.

Для обчислення НСД використовується **алгоритм Евкліда**, який базується на співвідношенні $\operatorname{НСД}(a, b) = \operatorname{НСД}(b, a \bmod b)$:

```python
def nsd_search(a: int, b: int) -> int:
    while b != 0:
        a, b = b, a % b
    return a

```

### Найменше спільне кратне (НСК)

Найменшим спільним кратним двох цілих чисел називається найменше натуральне число, яке ділиться без остачі на кожне з них.

Завдяки зв'язку між НСД та НСК:

$$\operatorname{НСД}(a, b) \cdot \operatorname{НСК}(a, b) = a \cdot b$$

знаходження НСК зводиться до ділення добутку чисел на їхній НСД:

```python
def nsk_search(a: int, b: int) -> int:
    return (a * b) // nsd_search(a, b)

```
