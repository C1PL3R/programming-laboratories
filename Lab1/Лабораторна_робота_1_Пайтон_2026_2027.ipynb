# Лабораторна робота 1

## Теоретична частина

### Лінійні алгоритми

Алгоритм — це скінченна послідовність чітко визначених дій, виконання яких приводить до отримання необхідного результату. Одним із найпростіших типів алгоритмів є **лінійний алгоритм**. У ньому всі команди виконуються послідовно, одна за одною: ```введення даних -> обчислення -> формування результату -> виведення```. У Python така послідовність задається порядком розташування інструкцій у програмі:

```python
a = 15
b = 4
c = a + b
print(c)
```

Спочатку створюються змінні `a` та `b`, потім обчислюється `c`, після чого результат виводиться на екран. У межах цієї лабораторної роботи основна увага приділяється саме **послідовним алгоритмам**, у яких не використовуються цикли та оператори галуження.

### Змінні, типи даних, введення

Змінна — це ім'я, пов'язане з певним значенням, яке зберігається під час виконання програми. У Python змінна створюється за допомогою оператора присвоювання `=`:

```python
number = 127
mass = 2.5
name = "Python"
```

Python автоматично визначає тип значення. Найважливіші типи для нашого класу задач:

* `int` — цілі числа;
* `float` — числа з плаваючою крапкою;
* `str` — рядки;
* `bool` — логічні значення.

Тип змінної можна перевірити функцією `type()`:

```python
x = 25
print(type(x))   # <class 'int'>
```

При присвоєнні значення змінній чи при визначенні функції  можна явно вказати **анотації типів** — очікувані типи параметрів та результату:

```python
number: int = 25
    ...
```

```python
def calculate(number: int) -> str:
    ...
```

Тут `number: int` означає, що параметр передбачається як ціле число, а `-> str` — що функція повертає рядок. Анотації допомагають читати й аналізувати програму, але сам Python не забороняє передати значення іншого типу — це підказка, а не примусова перевірка.

Для введення інформації з клавіатури використовується `input()`. За замовчуванням `input()` завжди повертає значення типу `str`, навіть якщо користувач ввів число: після введення `973` змінна міститиме рядок `"973"`, а не ціле число. Тому для роботи з числовими даними потрібне явне перетворення типу:

```python
number = int(input("Enter number: "))     # перетворення в ціле
x = float(input("Enter x: "))             # перетворення в дійсне
```

Тип перетворення (`int` чи `float`) має відповідати математичному змісту величини.

### Арифметичні операції

Python підтримує основні арифметичні операції:

| Операція | Позначення | Приклад | Результат |
|---|---:|---:|---:|
| Додавання | `+` | `7 + 3` | `10` |
| Віднімання | `-` | `7 - 3` | `4` |
| Множення | `*` | `7 * 3` | `21` |
| Ділення | `/` | `7 / 3` | `2.333...` |
| Цілочисельне ділення | `//` | `7 // 3` | `2` |
| Остача | `%` | `7 % 3` | `1` |
| Піднесення до степеня | `**` | `2 ** 3` | `8` |

**Порядок виконання операцій:** дужки → піднесення до степеня `**` → множення, ділення, `//`, `%` → додавання й віднімання. Наприклад, `2 + 3 * 4` дає `14`, а не `20`, бо множення виконується раніше. Для явного визначення порядку варто використовувати дужки: `(2 + 3) * 4`.

**Дійсне ділення `/`** завжди повертає результат типу `float`, навіть якщо він математично цілий: `8 / 2` дає `4.0`, а не `4`. Використовується, коли потрібен дійсний результат.

**Цілочисельне ділення `//`** визначає цілу частину результату ділення: `17 // 5` дає `3` (оскільки $17 = 3\cdot5 + 2$), `125 // 10` дає `12`. Ця операція корисна, коли потрібно визначити, скільки разів одна ціла величина повністю міститься в іншій.

**Остача `%`** визначає остачу від цілочисельного ділення: `17 % 5` дає `2`; для числа, яке ділиться без остачі, `20 % 5` дає `0`. Операції `//` та `%` пов'язані співвідношенням $a=(a//b)\cdot b+(a\bmod b)$, що в Python можна записати як:

```python
a = (a // b) * b + (a % b)
```

Ця властивість фундаментальна для багатьох алгоритмів обробки цілих чисел, зокрема для розкладання числа на цифри чи частини без перетворення на рядок:

```python
c = n % 10      # остання цифра
bc = n % 100    # останні дві цифри
n // 10         # число без останньої цифри
```

Загальна схема **послідовного вилучення частин величини** (наприклад, переведення секунд у хвилини й секунди):

```python
part_1 = value // divisor_1
value = value % divisor_1

part_2 = value // divisor_2
value = value % divisor_2
```

Важливо не переставляти порядок цих двох рядків: якщо спочатку оновити `value`, вихідне значення для обчислення `part` буде втрачено. Зміна значення змінної тут — свідома частина алгоритму: поточне значення представляє залишок, який обробляється на наступному етапі.

**Складені оператори присвоювання** дозволяють скорочено записувати операції зміни змінної: `x = x + 5` можна записати як `x += 5`; аналогічно `-=`, `*=`, `/=`, `//=`, `%=`. Наприклад, `value %= 60` еквівалентно `value = value % 60`.

**Абсолютне значення** обчислюється функцією `abs()`: `abs(-15)` дає `15`. Вона дозволяє отримати невід'ємний результат без самостійної перевірки знака, наприклад `difference = abs(a - b)` — незалежно від того, яке з чисел більше.

### Округлення, форматування, f-рядки

Функція `round(x, n)` виконує числове округлення до `n` знаків після крапки: `round(3.1415926, 3)` дає `3.142`. Це змінює саме числове представлення результату.

Для **форматування результату під час виведення** зручніше використовувати форматування рядків — воно визначає, як значення буде показане користувачу, не змінюючи саме число:

```python
x = 12.34567
print(f"{x:.1f}")   # 12.3
print(f"{x:.2f}")   # 12.35
print(f"{x:.3f}")   # 12.346
```

Позначення `:.3f` означає: `f` — число з плаваючою крапкою, `3` — кількість знаків після крапки.

Форматування застосовне й до цілих чисел. Специфікатор `:02d` означає: `d` — десяткове ціле число, `2` — мінімальна ширина поля, `0` — заповнення зліва нулями:

```python
print(f"{7:02d}")   # 07
hours, minutes, seconds = 3, 7, 5
print(f"{hours:02d}:{minutes:02d}:{seconds:02d}")   # 03:07:05
```

Це приклад форматування даних без зміни їхнього числового значення.

**F-рядки** (`f-strings`) — спосіб формування рядків, у які можна безпосередньо вставляти значення змінних та вирази: `f"текст {вираз}"`. Усередині `{}` можна писати не лише змінну, а й вираз (`f"{a + b}"`, `f"{number % 10}"`) або специфікатор формату (`f"{value:.3f}"`, `f"{number:02d}"`). F-рядки особливо зручні для формування структурованого текстового результату.

### Функції

Функція — це іменований блок програми, який виконує певну операцію та за необхідності повертає результат:

```python
def function_name(parameter):
    # тіло функції
    return result
```

Наприклад:

```python
def square(number: int) -> int:
    return number ** 2

result = square(5)   # result = 25
```

**Параметр** — змінна, зазначена у визначенні функції (`number`); **аргумент** — конкретне значення, передане під час виклику (`7` у `square(7)`). Функція може мати декілька параметрів (`def calculate(a, b, c)`), і при позиційній передачі порядок аргументів має значення.

**Оператор `return`** завершує виконання функції та повертає значення — число, рядок або результат складного виразу:

```python
def calculate(a, b):
    return f"{a + b}"
```

Якщо функція повинна сформувати текстовий результат, усі обчислення виконують усередині неї, а наприкінці формують один рядок і повертають його через `return`.

Функції дозволяють розділяти програму на логічні частини, повторно використовувати код, ізолювати окремі обчислення, спрощувати тестування й робити програму зрозумілішою.

**Docstring** — рядок документації одразу після заголовка функції, що описує її призначення, параметри, результат і може містити приклади використання:

```python
def square(number: int) -> int:
    """
    Returns the square of a number.
    """
    return number ** 2
```

### Логіка без операторів галуження: `min()` і `max()`

`min()` повертає найменше з переданих значень, `max()` — найбільше:

```python
min(7, 12)              # 7
min(4, 8, 2, 15)        # 2
```

`min()` особливо корисна, коли потрібно вибрати величину, яка не повинна перевищувати певне обмеження, наприклад `quantity = min(available, required)` — узяти менше з кількості доступних елементів і необхідної кількості.

Це дозволяє обмежити величину **без оператора `if`**. Щоб обмежити `x` значенням `maximum`:

```python
x = min(x, maximum)
```

а щоб гарантувати нижню межу:

```python
x = max(x, minimum)
```

У загальному вигляді: $x_{\text{used}}=\min(x_{\text{required}},x_{\text{available}})$. Це важливий прийом для побудови **нерозгалужених (branch-free) алгоритмів**, коли результат автоматично задовольняє потрібну межу.

Загальніше: у деяких випадках умовний вибір можна замінити не лише `min()`/`max()`, а й математичними операціями, властивостями цілочисельного ділення й остачі, формулами. Обмеження на використання певних конструкцій (наприклад, заборона `if`) не означає, що математичну операцію неможливо виконати — означає лише, що треба знайти інший допустимий спосіб її реалізації.

### Рядки `str`

Рядок — це послідовність символів, наприклад `text = "Python"`. Рядки можна об'єднувати оператором `+`: `"Hello, " + name`. Проте для формування складніших результатів зручніше використовувати f-рядки: `f"{name} version {version}"`.

Рядок може містити символ переходу на новий рядок `"\n"`, наприклад `"First line\nSecond line"`. Це особливо важливо, коли функція повинна повернути один рядок, що містить кілька рядків текстового результату.

### `doctest` і тестування

`doctest` — стандартний модуль Python для автоматичної перевірки прикладів, записаних у документації. Приклади записуються у docstring після символів `>>>`:

```python
import doctest

def add(a: int, b: int) -> int:
    """
    Adds two numbers.

    >>> add(2, 3)
    5

    >>> add(10, 7)
    17
    """
    return a + b

doctest.testmod()
```

Python виконає наведені приклади й порівняє результат із зазначеним у docstring; якщо всі приклади правильні, `doctest` не повідомляє про помилки. Запустити перевірку можна й з консолі: `python -m doctest program.py -v`.

Хороший набір тестів перевіряє не лише один типовий приклад, а й: звичайні випадки, граничні та нульові значення, мінімальні допустимі значення, значення на межі зміни результату. Наприклад, для програми з цілими числами варто окремо перевірити: остачу, що дорівнює нулю, і ненульову остачу; результат із початковими нулями під час форматування; випадки, коли кілька проміжних величин дорівнюють нулю.

**Граничний випадок** — значення на межі допустимої області або з особливою математичною властивістю. Приклади: `0` для невід'ємної величини; `100` і `999` — межі тризначного числа; `0, 1, 9, 10, 59` — межі двозначного формату. Саме на межах найчастіше виникають помилки в алгоритмах, тому граничні випадки мають особливе значення під час тестування.

Якщо `if`, `elif`, `else` заборонені, перевірку деяких обмежень реалізують на рівні формул: `value = min(value, limit)` — верхня межа, `value = max(value, minimum)` — нижня межа. Це не замінює всі можливі умовні конструкції, але для задач, побудованих на числових обмеженнях, зазвичай достатньо.

Іноді умову не обов'язково перевіряти явно — достатньо обчислити величину, яка сама характеризує стан системи. Наприклад, якщо після послідовних операцій залишилася величина `remaining`, то `remaining == 0` означає, що вся величина оброблена, а `remaining > 0` — що залишилась необроблена частина. Так можна вивести діагностичне значення, не створюючи окремої конструкції `if` — це приклад розділення обчислення стану та інтерпретації стану.

### Реалізація вибору результату без операторів галуження

У Python певні задачі, які зазвичай розв'язуються за допомогою операторів `if`, `elif` та `else`, можна реалізувати без явного галуження, використовуючи властивості булевих значень та арифметичні операції. У Python значення `True` і `False` під час числових операцій еквівалентні відповідно до `1` і `0`. Для явного перетворення логічного результату на ціле число можна використати функцію `int()`:

```python
int(True)   # 1
int(False)  # 0
```

Цю властивість можна використати для керування результатом обчислення. Наприклад, вираз

```python
valid = int(100 <= number <= 999)
```

присвоює `valid` значення `1`, якщо `number` є трицифровим додатним числом, і `0` в іншому випадку.

Додатково в Python рядок можна помножити на ціле число. Множення на `1` залишає рядок без змін, а множення на `0` дає порожній рядок:

```python
"Correct" * 1   # "Correct"
"Correct" * 0   # ""
```

Завдяки цьому декілька можливих результатів можна сформувати арифметично. Наприклад:

```python
negative = int(number <= 0)
valid = int(100 <= number <= 999)

result = (
    "Number must be positive" * negative
    + "Number of digits must be 3" * (1 - negative) * (1 - valid)
    + "Correct number" * valid
)
```

Для від'ємного числа `negative = 1`, тому перший рядок включається до результату, а інші множаться на `0`. Для додатного числа, яке не є трицифровим, `negative = 0` і `valid = 0`, тому формується повідомлення про неправильну кількість цифр. Для правильного трицифрового числа `valid = 1`, тому формується основний результат. Такий підхід дозволяє реалізувати простий вибір результату **без операторів галуження, циклів, списків, кортежів і словників**, використовуючи лише логічні та арифметичні операції. Водночас у звичайному прикладному програмуванні оператори `if` зазвичай є значно зрозумілішими; наведений прийом доцільний насамперед як навчальна демонстрація можливостей базових конструкцій Python.

### Стиль програмування та PEP 8

Під час написання програм важливо дотримуватися не лише правильності алгоритму, а й **єдиного стилю оформлення коду**. Основний документ, що описує рекомендований стиль для Python, — **PEP 8 (Style Guide for Python Code)**. Він містить рекомендації щодо відступів, довжини рядків, пробілів, назв змінних і функцій, оформлення імпортів та структури програм. Дотримання цих правил особливо важливе під час командної розробки, коли код читають і змінюють різні програмісти.

**Відступи.** Рекомендовано 4 пробіли; табуляцію не слід змішувати з пробілами. Кожен наступний рівень вкладеності додає ще 4 пробіли.

**Пробіли навколо операторів.** Навколо бінарних операторів варто ставити пробіли: `total = price + tax`, а не `total=price+tax`. Усередині дужок зайві пробіли не потрібні: `calculate(x, y)`, а не `calculate( x, y )`.

**Іменування змінних і функцій.** Використовується стиль **`snake_case`**: слова пишуться малими літерами, розділені `_` (`purchase_price`, `calculate_payment()`). Назва повинна якомога точніше описувати дані чи дію: краще `remaining_amount`, ніж `q`. Водночас назва не повинна бути надмірно довгою — краще `elapsed_seconds`, ніж `number_of_seconds_that_have_passed_since_the_beginning_of_the_day`. Назви не можуть починатися з цифри, містити пробіли чи збігатися із зарезервованими словами (`if`, `for`, `while`, `class`, `return` тощо); варто уникати назв, що відрізняються лише однією літерою.

### Побудова нерозгалужених лінійних алгоритмів

Для задач, у яких заборонено оператори галуження та цикли, корисно дотримуватися такої схеми:

1. Отримати початкові дані.
2. Перетворити їх до необхідних типів.
3. Виконати послідовність арифметичних операцій.
4. Використати `//` та `%`, якщо потрібно розкласти цілу величину на частини.
5. Використати `min()` або `max()`, якщо потрібно обмежити результат.
6. Послідовно оновлювати проміжні значення.
7. Сформувати результат за допомогою f-рядка.
8. Повернути результат із функції.
9. Перевірити функцію за допомогою `doctest`.

Основна ідея — не описувати програмно всі можливі випадки окремими гілками, а побудувати математичну послідовність операцій, яка автоматично працює для всієї допустимої області вхідних даних.

**Обмеження конструкцій у цій лабораторній роботі.** Заборонено оператори галуження (`if`, `elif`, `else`) та цикли (`for`, `while`). Дозволені: арифметичні операції `+ - * / // % **`, а також функції `abs()`, `min()`, `max()`, `round()`, `int()`, `float()`, `input()`, `print()` та f-рядки. Основний акцент — на арифметиці, цілочисельному діленні, остачі, функціях і форматуванні, а не на керуванні потоком виконання програми.

### Типові помилки

**Неправильне використання `/` замість `//`.** `125 / 10` дає `12.5`; якщо потрібна ціла кількість повних частин, слід використовувати `125 // 10`.

**Неправильне тлумачення `%`.** `125 % 10` дає `5` — це остання цифра числа, а не результат ділення. Оператор `%` повертає саме залишок.

**Втрата проміжного залишку.** Якщо значення послідовно розкладається на частини, важливо в правильному порядку оновлювати залишок: спочатку обчислити `part = value // divisor`, а потім `value = value % divisor`. Зміна порядку призводить до втрати початкових даних.

**Неправильне форматування.** Значення `5` і рядок `"05"` математично те саме число, але текстово різні; для отримання `05` потрібне `f"{5:02d}"`.

**Змішування числових та текстових даних.** `input()` повертає рядок, тому арифметичні операції над ним без перетворення (`x = int(input())`) можуть бути некоректними.

**Змішування логіки та виведення.** Функція, що виконує обчислення, повинна повертати результат (`return result`), а вже поза функцією його можна виводити (`print(result)`). Це спрощує тестування та повторне використання коду.

Користувач здійснює покупку в магазині. Він повідомляє **суму покупки** та кількість наявних банкнот кожного номіналу. Доступні банкноти мають номінали: $500; 100; 50; 20; 10; 5; 1$ (у гривнях).

Необхідно написати програму, яка визначає, **скільки банкнот кожного номіналу потрібно використати для оплати покупки**, починаючи з найбільшого номіналу. При цьому кількість доступних банкнот кожного номіналу є обмеженою. Для кожного номіналу необхідно використати максимально можливу кількість банкнот, яка одночасно:

* не перевищує поточну суму, яку ще потрібно сплатити;
* не перевищує кількість доступних банкнот цього номіналу.

Для визначення кількості банкнот дозволяється використовувати функцію `min()`.

#### Вхідні дані

Користувач послідовно вводить:

1. суму покупки `price`;

4. кількість банкнот номіналом 50 грн;

7. кількість банкнот номіналом 5 грн;
8. кількість банкнот номіналом 1 грн.

Усі значення є **невід'ємними цілими числами**.
