# Теоретичні відомості

## 1. Алгоритмічний підхід до розв’язування задач

Більшість прикладних задач, які розв’язуються за допомогою програмування, можна поділити на кілька етапів. Спочатку необхідно проаналізувати умову задачі, визначити початкові дані та бажаний результат. Далі встановлюються правила або математичні залежності, за якими з початкових даних отримують результат.

У загальному випадку алгоритм можна подати у вигляді схеми: 
$\text{вхідні дані}
\longrightarrow
\text{обробка}
\longrightarrow
\text{результат}.
$

Обробка даних може містити арифметичні операції, порівняння, вибір одного з декількох варіантів, повторення певних дій або звернення до допоміжних алгоритмів. Важливо, що алгоритм повинен враховувати не тільки звичайні випадки, а й граничні та некоректні значення. Саме аналіз таких випадків часто визначає правильність розв’язання задачі.

## 2. Дискретні та числові дані

У багатьох задачах вхідні дані є цілими числами. Це можуть бути кількість років, кількість елементів, номер певного об’єкта, кількість одиниць товару, час у секундах тощо. Для цілих чисел важливими є не лише самі значення, а й певні властивості чисел: парність, остання цифра, кількість десятків, належність до певного діапазону. Наприклад, парність числа $n$ визначається його залишком при діленні на $2$:

$$
n\bmod 2=
\begin{cases}
0,& n\text{ парне},\\
1,& n\text{ непарне}.
\end{cases}
$$

Операція взяття залишку від ділення є важливим інструментом під час роботи з дискретними даними.

## 3. Арифметичні властивості десяткового запису числа

Залишок від ділення на степені числа $10$ дозволяє аналізувати окремі цифри десяткового запису. Остання цифра числа визначається як
$n\bmod 10$. Останні дві цифри можна отримати за допомогою $n\bmod 100$.

Наприклад, якщо потрібно визначити, чи належить число до певної категорії залежно від останньої цифри, немає потреби перетворювати число на текст. Достатньо використати його арифметичні властивості. Цей підхід особливо корисний у задачах, де результат залежить від десяткового запису числа. Наприклад, правила утворення форм слів після числівників можна описати через останню цифру та останні дві цифри числа.

Для чисел української мови, які визначають форму слова після кількості років, важливими є такі закономірності:

* числа, що закінчуються на $1$, за винятком $11$, відповідають формі «рік»;
* числа, що закінчуються на $2$, $3$ або $4$, за винятком $12$, $13$, $14$, відповідають формі «роки»;
* інші випадки відповідають формі «років».

Тому для правильної класифікації недостатньо аналізувати лише останню цифру. Необхідно окремо враховувати числа від $11$ до $14$.

Це є прикладом задачі, у якій **формальна логіка та властивості чисел дозволяють отримати результат без складних математичних обчислень**.

## 4. Діапазони значень та граничні випадки

У практичних задачах результат часто залежить від того, до якого діапазону належить величина. Наприклад, певна величина $x$ може мати три області:

$$
x< a,
$$

$$
a\leq x\leq b,
$$

$$
x>b.
$$

Для кожної області може бути визначене своє правило обробки. Особливу увагу потрібно приділяти граничним значенням $a$ і $b$. Умова $
x<a$ відрізняється від $x\leq a$, оскільки в першому випадку значення $x=a$ не належить відповідній області, а в другому — належить. Під час розроблення алгоритму необхідно чітко визначити, як обробляються:

* мінімально допустимі значення;
* максимально допустимі значення;
* значення безпосередньо біля межі;
* значення поза допустимим діапазоном.


## 5. Умовні алгоритми

Якщо результат залежить від значення певної величини, використовується **розгалуження алгоритму**. У найпростішому випадку існують два можливі варіанти:

$$
f(x)=
\begin{cases}
f_1(x), & \text{якщо виконується умова},\\
f_2(x), & \text{в іншому випадку}.
\end{cases}
$$

Якщо можливих варіантів більше, використовують послідовність умов:

$$
\begin{cases}
f_1(x), & A_1,\\
f_2(x), & A_2,\\
f_3(x), & A_3.
\end{cases}
$$

Такі алгоритми називають **розгалуженими**. У програмуванні розгалуження дозволяє безпосередньо реалізувати логіку математичної моделі або правила, заданого умовою задачі.


## 6. Логічні умови

Умови можуть бути простими або складеними. Проста умова порівнює дві величини: $x>0$, $x=y$, $n\leq 140$.


Складна умова може містити декілька тверджень. Наприклад, належність числа до інтервалу можна записати як $a\leq x\leq b$.

У програмуванні складні умови будуються за допомогою логічних операцій. Основними логічними операціями є:

* **AND** — одночасне виконання умов;
* **OR** — виконання хоча б однієї умови;
* **NOT** — заперечення умови.

Логічні умови дають змогу формально описувати правила, які в текстовій формі можуть здаватися складними.


## 7. Перевірка коректності даних

Перед виконанням основного алгоритму часто необхідно перевірити, чи мають вхідні дані допустимий зміст. Наприклад, якщо деяка величина повинна бути невід’ємною, то допустима область визначається умовою $x\geq0$. Якщо величина повинна належати певному інтервалу, $a\leq x\leq b$. Перевірка коректності даних є важливою частиною алгоритму, оскільки математична формула сама по собі не гарантує фізичної або логічної коректності результату. Наприклад, комп’ютер може без проблем виконати арифметичні операції з числом, яке не має сенсу в конкретній задачі. Тому **математична можливість виконати обчислення не завжди означає коректність задачі**.


## 8. Функції як спосіб організації алгоритму

Під час розв’язування задач доцільно виділяти окремі завершені операції в самостійні функції. Функція може розглядатися як перетворення $
f:X\rightarrow Y$, яке кожному допустимому набору вхідних значень ставить у відповідність певний результат.

Наприклад,

$$
f(n)=
\begin{cases}
\text{«рік»}, & \text{за певних умов},\\
\text{«роки»}, & \text{за інших умов},\\
\text{«років»}, & \text{в інших випадках}.
\end{cases}
$$

Перевага такого підходу полягає в тому, що окреме правило можна визначити один раз, а потім використовувати його багаторазово. Функції також полегшують перевірку програми: кожну частину алгоритму можна тестувати незалежно.

## 9. Декомпозиція та композиція функцій

Складну задачу зручно розділяти на простіші логічні компоненти. Такий процес називається **декомпозицією**. Наприклад, загальну задачу можна представити як $F(x)=G(H(x))$, де спочатку виконується $H(x)$, а отриманий результат передається функції $G$. У програмуванні це означає, що одна функція може викликати іншу. Такий підхід має кілька переваг:

* зменшує складність окремих частин програми;
* дозволяє повторно використовувати вже створені функції;
* спрощує тестування;
* робить алгоритм зрозумілішим;
* дозволяє відокремити загальні правила від конкретної задачі.

Отже, функція може бути не лише кінцевою процедурою обчислення, а й **будівельним блоком складнішого алгоритму**.


## 10. Повторне використання алгоритмів

Якщо певне правило використовується в декількох місцях програми, його доцільно реалізувати один раз. Наприклад, якщо для різних значень потрібно визначати одну й ту саму характеристику числа, замість повторення однакової послідовності перевірок створюють функцію. Цей принцип називають **повторним використанням коду**. Він є важливим не лише для невеликих навчальних програм. У великих програмних системах повторне використання функцій, модулів та бібліотек є одним з основних способів зменшення складності програмного забезпечення.

## 11. Тестування алгоритмів

Коректність алгоритму необхідно перевіряти на різних наборах даних. Для цього доцільно використовувати:

### Звичайні випадки

Перевіряються типові значення, для яких результат легко передбачити.

### Граничні випадки

Перевіряються значення безпосередньо на межах допустимих областей.

### Спеціальні випадки

Перевіряються значення, які мають особливі математичні або логічні властивості.

### Некоректні випадки

Перевіряється реакція алгоритму на дані, які не відповідають умовам задачі.

Такий поділ тестів дозволяє перевірити не лише основну логіку, а й поведінку програми в нетипових ситуаціях.

## 12. Документування та автоматична перевірка функцій

Функція повинна мати чітко визначені:

* призначення;
* вхідні параметри;
* допустимі значення;
* результат;
* особливі випадки.

У Python приклади правильної роботи функції можна записувати в її документації та перевіряти автоматично за допомогою механізму `doctest`. Такий підхід поєднує документацію і тестування. Приклад використання функції одночасно пояснює її призначення та задає очікуваний результат.
Особливо корисно включати до тестів граничні та спеціальні випадки.

# 13. Математичне моделювання фізичних процесів

Розглянуті вище принципи можна застосовувати не лише до дискретних задач, а й до моделювання фізичних явищ. У цьому випадку алгоритм будується на основі фізичного закону або математичного рівняння. Типова послідовність має вигляд

$$
\text{фізичне явище}
\rightarrow
\text{фізична модель}
\rightarrow
\text{математичні рівняння}
\rightarrow
\text{чисельний алгоритм}.
$$

Таким чином, програмування стає інструментом дослідження математичної моделі фізичної системи.

## 14. Опис механічного руху

Положення матеріальної точки можна описати радіус-вектором $\vec r=\vec r(t)$. Швидкість визначається як $\vec v=\frac{d\vec r}{dt}$, а прискорення — як $\vec a=\frac{d\vec v}{dt}$. Для руху з постійним прискоренням $\vec a=\mathrm{const}$ маємо $\vec v(t)=\vec v_0+\vec a t$ та $\vec r(t)=\vec r_0+\vec v_0t+\frac{\vec a t^2}{2}$. Ці співвідношення дозволяють перетворити фізичну задачу про рух на систему числових обчислень.


## 15. Рух у двовимірному просторі

Якщо тіло рухається в площині, його положення задається двома координатами: $\vec r(t)=(x(t),y(t))$. Відповідно, рух можна розглядати окремо вздовж двох координатних напрямків. Для кожної координати записується власне рівняння руху: $x=x(t)$, $y=y(t)$.

Такий підхід є прикладом **декомпозиції фізичної задачі**: складний двовимірний рух розглядається як сукупність простіших рухів уздовж окремих осей.


## 16. Рух тіла в полі тяжіння

Поблизу поверхні Землі, якщо опором повітря нехтувати, прискорення тіла є сталим і напрямленим вертикально вниз:

$$
\vec a=(0,-g).
$$

Тому горизонтальна складова швидкості залишається сталою:

$$
v_x=v_{0x},
$$

а вертикальна змінюється за законом

$$
v_y=v_{0y}-gt.
$$

Координати тіла визначаються співвідношеннями

$$
x(t)=x_0+v_{0x}t,
$$

$$
y(t)=y_0+v_{0y}t-\frac{gt^2}{2}.
$$

Ці рівняння є математичною моделлю руху тіла, кинутого під кутом до горизонту.



## 17. Векторне представлення початкової швидкості

Якщо тіло запускається зі швидкістю \(v_0\) під кутом \(\alpha\) до горизонталі, початкову швидкість можна розкласти на компоненти:

$$
v_{0x}=v_0\cos\alpha,
$$

$$
v_{0y}=v_0\sin\alpha.
$$

Тоді горизонтальний та вертикальний рух описуються незалежними рівняннями.

Модуль швидкості в будь-який момент часу визначається через її компоненти:

$$
v(t)=\sqrt{v_x^2(t)+v_y^2(t)}.
$$

Цей принцип — розкладання вектора на компоненти — широко використовується у фізиці та чисельному моделюванні.


## 18. Аналіз траєкторії та екстремальних значень

Знання рівнянь руху дозволяє визначати не лише положення тіла в певний момент часу, а й характеристики всієї траєкторії. Наприклад, максимальна висота досягається в момент, коли вертикальна складова швидкості дорівнює нулю:

$$
v_y=0.
$$

Звідси

$$
t_{\max}=\frac{v_{0y}}{g}.
$$

Максимальна висота становить

$$
H_{\max}
=
H+\frac{v_{0y}^2}{2g}.
$$

Підставляючи початкову вертикальну складову швидкості, отримуємо

$$
H_{\max}
=
H+\frac{v_0^2\sin^2\alpha}{2g}.
$$

Подібним чином можна визначити час падіння та горизонтальну дальність польоту.

Це демонструє важливий принцип математичного моделювання: **одна модель може використовуватися для отримання різних характеристик досліджуваного процесу**.


## 19. Перетворення одиниць у математичній моделі

Під час роботи з фізичними величинами необхідно дотримуватися узгоджених одиниць вимірювання.

У механіці зазвичай використовують систему SI:

* довжина — метр;
* час — секунда;
* швидкість — метр за секунду;
* прискорення — метр за секунду в квадраті.

Особливу увагу потрібно приділяти кутам. У математичних функціях `sin` і `cos` аргумент зазвичай задається в **радіанах**, тоді як у фізичній задачі кут часто задається в градусах.

Перехід від градусів до радіанів здійснюється за формулою

$$
\alpha_{\mathrm{rad}}
=
\alpha_{\mathrm{deg}}\frac{\pi}{180}.
$$

Тому правильне врахування одиниць є необхідною частиною побудови фізичної математичної моделі.


## 20. Від алгоритму до обчислювальної моделі

Таким чином, задачі різного типу — від обробки цілих чисел до моделювання механічного руху — мають спільну основу. У дискретних задачах алгоритм може визначати, до якої категорії належить число:

$$
x\rightarrow\text{категорія}.
$$

У фізичній задачі алгоритм може обчислювати значення фізичної величини:

$$
(t,H,v_0,\alpha)
\rightarrow
(x,y,v).
$$

В обох випадках необхідно:

1. визначити вхідні параметри;
2. встановити закономірності між ними;
3. визначити допустиму область значень;
4. розділити складну задачу на логічні компоненти;
5. реалізувати компоненти у вигляді функцій;
6. врахувати граничні та спеціальні випадки;
7. перевірити правильність результатів.

Отже, основною метою програмування в таких задачах є не механічне виконання окремих команд, а **формалізація закономірностей і побудова алгоритмів, здатних застосовувати ці закономірності до різних наборів даних**.

## 22. Модуль mypy
